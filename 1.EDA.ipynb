### Verificação de valores ausentes e duplicados

Nesta etapa, foi avaliada a qualidade inicial da tabela `orders` por meio da identificação de valores ausentes em cada coluna.

Para cada coluna, foi calculado:

- **Quantidade de valores nulos:** número de registros sem informação.
- **Percentual de valores nulos:** proporção de registros sem informação em relação ao total de linhas.

A fórmula utilizada foi:

\[
\text{Percentual de nulos} =
\frac{\text{Quantidade de valores nulos na coluna}}
{\text{Quantidade total de linhas}}
\times 100
\]

No código, `isnull()` identifica valores ausentes. Como valores booleanos podem ser interpretados como 1 para `True` e 0 para `False`, a função `mean()` calcula diretamente a proporção de nulos na coluna. Por fim, os resultados foram ordenados de forma decrescente para priorizar as colunas com maior ausência de informação.

A análise permite identificar campos que exigem atenção em etapas posteriores, sem realizar qualquer tratamento ou remoção de dados nesta fase.

````python
nulos = orders.isnull().sum().sort_values(ascending=False)
percentual_nulos = (orders.isnull().mean() * 100).sort_values(ascending=False)

qualidade = pd.DataFrame({
    "valores_nulos": nulos,
    "percentual_nulo": percentual_nulos.round(2)
})

qualidade
````

## 3.1 Análise de outliers pelo método IQR

Para aprofundar a análise da coluna `total`, foi utilizado o método do Intervalo Interquartil (IQR).

O IQR corresponde à diferença entre o terceiro quartil (Q3) e o primeiro quartil (Q1). Foram considerados outliers os valores abaixo de `Q1 - 1,5 × IQR` ou acima de `Q3 + 1,5 × IQR`.

````python
## aplicação da fórmula de IQR 
q1 = orders["total"].quantile(0.25)
q3 = orders["total"].quantile(0.75)

iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

outliers = orders[
    (orders["total"] < limite_inferior) |
    (orders["total"] > limite_superior)
]

quantidade_outliers = len(outliers)
percentual_outliers = (quantidade_outliers / len(orders)) * 100

print(f"Q1: R$ {q1:,.2f}")
print(f"Q3: R$ {q3:,.2f}")
print(f"IQR: R$ {iqr:,.2f}")
print(f"Limite inferior: R$ {limite_inferior:,.2f}")
print(f"Limite superior: R$ {limite_superior:,.2f}")
print(f"\nQuantidade de outliers: {quantidade_outliers:,}")
print(f"Percentual da base: {percentual_outliers:.2f}%")
print(f"Menor outlier: R$ {outliers['total'].min():,.2f}")
print(f"Maior outlier: R$ {outliers['total'].max():,.2f}")
````

Foram identificados 452 outliers, equivalentes a 0,92% da base. O primeiro quartil foi de R$ 13.171,24, o terceiro quartil foi de R$ 40.941,88 e o IQR foi de R$ 27.770,65. Com isso, foi definido um limite superior de R$ 82.597,85.

Não foram identificados outliers inferiores, pois conforme informado anteriormente, o limite inferior calculado foi negativo (R$ -28.484,74). Todos os 452 outliers são pedidos de alto valor, entre R$ 82.618,00 e R$ 127.262,02.

## 3.2 -  Impacto financeiro dos registros atípicos

Após a identificação dos outliers pela metodologia IQR, precisa ser avaliado se esses pedidos representam apenas uma ocorrência estatística pontual ou se possuem impacto relevante no faturamento da empresa.

Para isso, serão calculados:

- o faturamento total da base;
- o faturamento associado aos pedidos classificados como outliers;
- a participação percentual desses registros no faturamento total;
- a média dos pedidos após a exclusão temporária dos outliers.

A participação dos outliers no faturamento será calculada pela fórmula:

\[
\text{Participação dos outliers} =
\frac{\text{Faturamento dos outliers}}
{\text{Faturamento total}}
\times 100
\]

A média sem outliers será utilizada apenas como análise de sensibilidade, para comparar o comportamento dos pedidos típicos com a média original. Nenhum registro será removido da base.

````python
# Soma o valor total de todos os pedidos da base.
faturamento_total = orders["total"].sum()

# Soma somente o valor dos pedidos identificados como outliers pelo IQR.
faturamento_outliers = outliers["total"].sum()

# Calcula o percentual do faturamento total concentrado nos outliers.
participacao_faturamento = (
    faturamento_outliers / faturamento_total
) * 100

# Remove temporariamente os índices presentes em outliers
# para calcular a média dos pedidos sem valores atípicos.
media_sem_outliers = orders.loc[
    ~orders.index.isin(outliers.index),
    "total"
].mean()

# Apresenta os resultados em formato monetário e percentual.
print(f"Faturamento total: R$ {faturamento_total:,.2f}")
print(f"Faturamento dos outliers: R$ {faturamento_outliers:,.2f}")
print(f"Participação no faturamento: {participacao_faturamento:.2f}%")
print(f"Média sem outliers: R$ {media_sem_outliers:,.2f}")
````

Apesar de representarem uma parcela pequena da quantidade de pedidos, esses registros concentram **2,94%** do faturamento total. A média geral é R$ 28.704,99, enquanto a média sem os outliers é R$ 28.119,80, demonstrando que os pedidos extremos elevam a média.
